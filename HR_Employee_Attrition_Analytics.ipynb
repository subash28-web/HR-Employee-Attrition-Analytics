## Attrition analysis
### Attrition by department

```python
query8="""select "Department" , COUNT(*) as total_employees,
            SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
            ROUND(SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),2) AS attrition_rate 
            from employee
            group by "Department";"""

attri_by_dept=pd.read_sql(query8,engine)
attri_by_dept
```

### Attrition by age group

```python
query17="""select CASE 
                    WHEN "Age" < 25 THEN '18-25'
                    WHEN "Age" <=35 THEN '25-35'
                    WHEN "Age" <=45 THEN '36-45'
                    WHEN "Age" <=55 THEN '46-55'
                    ELSE '56+'
                END as age_group,
                COUNT(*) as total_employees,
                COUNT(*) FILTER (where "Attrition" = 'Yes') as employees_left,
                ROUND(COUNT(*) FILTER (where "Attrition" = 'Yes') * 100.0 / COUNT(*),2 ) as attrition_rate
                from employee
                group by age_group
                order by age_group ASC;"""

attri_by_age_group=pd.read_sql(query17,engine)
attri_by_age_group
```

### Attrition by Income Group

```python
query22="""select CASE
                    WHEN "MonthlyIncome" < 3000 THEN 'Below 3000'
                    WHEN "MonthlyIncome" <= 6000 THEN '3000-6000'
                    WHEN "MonthlyIncome" <= 10000 THEN '6000-10000' 
                    ELSE 'Above 10000'
                END as income_group,
                COUNT(*) as total_employees,
                SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
                ROUND(SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),2) AS attrition_rate 
            from employee
            group by "income_group"
            order by attrition_rate DESC;"""

attri_by_income_group=pd.read_sql(query22,engine)
attri_by_income_group
```

### Do employees with lower income leave more?

```python
query21="""select CASE
                    WHEN "MonthlyIncome" < 3000 THEN '-3k'
                    WHEN "MonthlyIncome" <= 6000 THEN '3k-6k'
                    WHEN "MonthlyIncome" <= 10000 THEN '6k-10k' 
                    ELSE '10k+'
                END as income_group,
                COUNT(*) as total_employees,
                SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
                ROUND(SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),2) AS attrition_rate 
            from employee
            group by "income_group"
            order by attrition_rate DESC
            limit 1;                 
                    """

attri_by_income_group=pd.read_sql(query21,engine)
attri_by_income_group
```

### Does distance from home affect attrition?

```python
query23="""select CASE
                    WHEN "DistanceFromHome" <= 5 THEN '0-5 km'
                    WHEN "DistanceFromHome" <= 10 THEN '6-10 km'
                    WHEN "DistanceFromHome" <= 20 THEN '11-20 km' 
                    ELSE '20 km+'
                END as distance_group,
                COUNT(*) as total_employees,
                SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
                ROUND(SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),2) AS attrition_rate 
            from employee
            group by "distance_group"
            order by attrition_rate DESC;"""

attri_by_distance_group=pd.read_sql(query23,engine)
attri_by_distance_group
```

### Does longer tenure reduce attrition?

```python
query24="""select CASE
                    WHEN "YearsAtCompany" <= 2 THEN '0-2 year'
                    WHEN "YearsAtCompany" <= 5 THEN '3-5 year'
                    WHEN "YearsAtCompany" <= 10 THEN '6-10 year' 
                    ELSE '10 year+'
                END as tenure_group,
                COUNT(*) as total_employees,
                SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
                ROUND(SUM(CASE WHEN "Attrition" = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),2) AS attrition_rate 
            from employee
            group by "tenure_group"
            order by attrition_rate DESC;"""

attri_by_income_group=pd.read_sql(query24,engine)
attri_by_income_group
```

### Which employee group is most at risk of leaving?

```python
query25="""select
                CASE 
                    WHEN "Age" < 30 THEN 'Below 30'
                    WHEN "Age" <=40 THEN '30-40'
                    WHEN "Age" <=50 THEN '41-50'
                    ELSE '55+'
                END as age_group,
                "OverTime",
                CASE
                    WHEN "MonthlyIncome" < 3000 THEN 'Low Income'
                    WHEN "MonthlyIncome" < 6000 THEN 'Medium Income'
                    ELSE 'High Income'
                END AS income_group,
                COUNT(*) as total_employees,
                COUNT(*) FILTER (where "Attrition" = 'Yes') as employees_left,
                ROUND(COUNT(*) FILTER (where "Attrition" = 'Yes') * 100.0 / COUNT(*),2 ) as attrition_rate
                from employee
                group by age_group,"OverTime",income_group
                having count(*) >= 20
                order by attrition_rate DESC;"""

most_risk=pd.read_sql(query25,engine)
most_risk
```
